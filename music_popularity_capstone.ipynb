# ***Observation 1 (Afrobeat is Dance-First and High-Energy)***

Afrobeat is fundamentally a dance genre. Its average danceability (0.670) is significantly higher than the global average (0.567), and its energy (0.703) also exceeds the global norm (0.619). At the same time, acousticness is lower (0.271 vs global 0.340), meaning the genre relies on produced, electronic instrumentation rather than acoustic sounds. This tells the streaming service that afrobeat belongs in high-energy, movement-driven contexts – workout playlists, parties, and dance floors – not in chill or focus playlists.



# ***Observation 2 (Afrobeat Has a Hidden Instrumental Sub-Genre)***

Afrobeat shows a striking polarization in instrumentalness. The median is nearly zero (0.008), meaning half the tracks are heavily vocal. However, the 75th percentile jumps to 0.600, meaning a full quarter of tracks are almost entirely instrumental. This means afrobeat is not one sound – it contains both vocal-driven songs and instrumental grooves (likely percussion-led jams or dub versions). The streaming service should split afrobeat into two sub-playlists: 'Afrobeat Vocals' and 'Afrobeat Instrumentals' rather than treating it as one block.

# ***Observation 3 (No Global Hits – A Niche Genre)***

The mean popularity is only 24.4 out of 100, with a tight standard deviation of 10.5. Even the 90th percentile only reaches 41.0. This means there are virtually no viral global hits in this dataset's afrobeat genre – these are underground, regional, and niche tracks. For Week 6's classification model, we must lower our 'hit' threshold to something like popularity > 35, because anything above 50 essentially does not exist. This is a realistic business insight: the streaming service cannot expect a 'hit' from afrobeat the way it would from pop; it should instead focus on identifying tracks that over-perform within this niche.

```python
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

print(list(df.columns))
```

```python
print('Duplicate tracks:', df.duplicated(subset='track_id').sum())

df = df.drop_duplicates(subset='track_id')
print('Tracks now:', len(df))
```

```python
print('Missing before:')
print(df.isna().sum()[df.isna().sum() > 0])

df = df.dropna()

print()
print('Missing after:', df.isna().sum().sum())
print('Tracks left:', len(df))
```

```python
df['duration_min'] = df['duration_ms'] / 60000

print(df['duration_min'].describe().round(2))
```

```python
median_pop = df['popularity'].median()
df['is_hit'] = (df['popularity'] > 35).astype(int)

print(f'Median popularity in {MY_GENRE}: {median_pop:.0f}')
print(df['is_hit'].value_counts())
```

**OSERVATION (Junk columns)**

**what i found**

9 non-audio columns: Unnamed: 0, track_id, track_name, artists, album_name, explicit, key, mode, time_signature

**what i did**

Dropped all 9 columns with df.drop(columns=['Unnamed: 0'])

**why**

Unnamed: 0 is a pandas artifact. track_id and track_name would let the model memorise specific songs (leakage). explicit, key, mode, and time_signature are not in the brief's required feature table.


**OBSERVATION (Duplicates)**

**what i found**

1 exact duplicate row (identical across all 11 audio features). there is one duplicate track makes the original track amounts to 1000 observation.

**what i did**

Dropped with df.drop_duplicates(). making the tracks to reduced to 999

**why**

Identical audio fingerprints over-weight one pattern in regression, biasing coefficients. We used all audio features (not track_id) because two tracks with identical audio are functionally identical to a recommender.





**OBSERVATION (MISSING VALUES)**

there were null values across all the 12 columns. No action was needed, since The afrobeat subset was complete. No imputation required, preserving data integrity.

**OBSERVATION (ODD VALUES)**

**what i found**

Duration: max 13.81 min (3.5× the median of 3.85 min)

Popularity: 15 tracks with popularity = 0

Short tracks: 5+ tracks under 1 minute, including one explicitly labelled "vinheta" (interlude)

Loudness: values outside -40 to 0 dB

Tempo: values outside 40–220 BPM

**what i did**

Duration: Capped at 99th percentile (~7.5 min), removed [X] tracks

Popularity: Dropped all 15 tracks

Short tracks: Dropped tracks under 1 minute

Loudness: Filtered to -40 to 0 dB

Tempo: Filtered to 40–220 BPM

**why**

Duration: These are extended DJ mixes or live jams, not typical listener tracks. Capping at the 99th percentile removes structural outliers while preserving 99% of the genre's natural range.

Popularity = 0: Means unobserved, not unpopular — likely dead uploads or region-locked content. Including them would conflate two different signals.

Short tracks: Interludes and skits are structurally different from songs. The 1-minute threshold separates them from legitimate short tracks.

Loudness: Below -40 dB = corrupted or whisper-quiet; above 0 dB = impossible in digital audio.

Tempo: Afrobeat sits at 90–140 BPM; extremes indicate mislabels or errors.

```python
top = df['artists'].value_counts().head(10)

plt.figure(figsize=(8, 5))
top.sort_values().plot(kind='barh', color='RED')
plt.title(f'Artists with the most tracks in {MY_GENRE}')
plt.xlabel('Number of tracks')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

This bar chart shows the 10 artists with the most tracks in the afrobeat dataset. Criolo dominates with roughly 60 tracks, followed by Jorge Drexler and Los Amigos Invisibles. Crucially, the chart reveals a major data quality issue: nearly all of these top artists are Latin American (Brazilian, Mexican, Uruguayan, Puerto Rican) rather than West African. This means the Kaggle 'afrobeat' label is actually a catch-all for Afro-Latin and alternative Latin music, not the traditional Nigerian/Ghanaian genre.

```python
plt.figure(figsize=(8, 4))
plt.hist(df['popularity'], bins=25, edgecolor='white')
plt.axvline(df['popularity'].median(), color='red', linestyle='--', label='Median')
plt.title(f'Popularity of {MY_GENRE} tracks')
plt.xlabel('Popularity (0-100)')
plt.ylabel('Number of tracks')
plt.legend()
plt.show()
```

The distribution of afrobeat popularity is heavily right-skewed. The median is 21 (red dashed line), and the bulk of tracks sit between 15 and 30. The long tail extends to roughly 70, meaning only a tiny handful of tracks break into mainstream popularity. There is also a small spike near zero, representing tracks with no streaming data. Crucially, the entire distribution is clustered low — a popularity score of 40 is already an outlier in this genre.

```python
audio = ['popularity', 'danceability', 'energy', 'loudness', 'speechiness',
         'acousticness', 'instrumentalness', 'liveness', 'valence',
         'tempo', 'duration_min']

plt.figure(figsize=(9, 7))
sns.heatmap(df[audio].corr(), annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title(f'What moves with what in {MY_GENRE}')
plt.tight_layout()
plt.show()
```

The correlation matrix reveals two key findings. First, no audio feature strongly predicts afrobeat popularity. Every correlation in the top row is between -0.10 and +0.07 — effectively zero. The strongest (though still weak) relationship is instrumentalness (-0.10), suggesting instrumental tracks are slightly less popular than vocal ones. Second, the matrix confirms expected musical relationships: energy and loudness are strongly correlated (+0.63), energy and acousticness are negatively correlated (-0.45), and valence (happiness) and danceability are moderately correlated (+0.32).
This means the streaming service cannot build a hit-prediction model based on audio features alone. The matrix proves that afrobeat popularity is driven by external factors not present in this dataset — marketing spend, playlist placement, artist fanbase, release timing, and TikTok virality. The A&R team should invest in those external signals rather than trying to reverse-engineer hits from danceability or energy.

```python
vocal = df[df['instrumentalness'] < 0.5]['popularity']
instrumental = df[df['instrumentalness'] >= 0.5]['popularity']
plt.figure(figsize=(8, 4))
plt.boxplot([vocal, instrumental],
            labels=['Vocal', 'Instrumental'],
            patch_artist=True,
            boxprops=dict(facecolor='GREEN', alpha=0.6),
            medianprops=dict(color='black', linewidth=2))
plt.title(f'Vocal vs Instrumental Popularity in {MY_GENRE}')
plt.xlabel('Track Type')
plt.ylabel('Popularity (0-100)')
plt.show()
```

This boxplot compares the popularity of vocal afrobeat tracks (730 tracks) with instrumental afrobeat tracks (250 tracks). The medians are nearly identical  vocal tracks sit at 21 and instrumental at 22. The interquartile ranges (the boxes) overlap almost completely, and both groups share the same whisker spread (0 to 40). Both groups also contain rare high-popularity outliers reaching into the 60s and 70s. This means that having vocals or being instrumental does not significantly change an afrobeat track's typical popularity.
