# ***Observation 1 (Afrobeat is Dance-First and High-Energy)***

Afrobeat is fundamentally a dance genre. Its average danceability (0.670) is significantly higher than the global average (0.567), and its energy (0.703) also exceeds the global norm (0.619). At the same time, acousticness is lower (0.271 vs global 0.340), meaning the genre relies on produced, electronic instrumentation rather than acoustic sounds. This tells the streaming service that afrobeat belongs in high-energy, movement-driven contexts – workout playlists, parties, and dance floors – not in chill or focus playlists.



# ***Observation 2 (Afrobeat Has a Hidden Instrumental Sub-Genre)***

Afrobeat shows a striking polarization in instrumentalness. The median is nearly zero (0.008), meaning half the tracks are heavily vocal. However, the 75th percentile jumps to 0.600, meaning a full quarter of tracks are almost entirely instrumental. This means afrobeat is not one sound – it contains both vocal-driven songs and instrumental grooves (likely percussion-led jams or dub versions). The streaming service should split afrobeat into two sub-playlists: 'Afrobeat Vocals' and 'Afrobeat Instrumentals' rather than treating it as one block.

# ***Observation 3 (No Global Hits – A Niche Genre)***

The mean popularity is only 24.4 out of 100, with a tight standard deviation of 10.5. Even the 90th percentile only reaches 41.0. This means there are virtually no viral global hits in this dataset's afrobeat genre – these are underground, regional, and niche tracks. For Week 6's classification model, we must lower our 'hit' threshold to something like popularity > 35, because anything above 50 essentially does not exist. This is a realistic business insight: the streaming service cannot expect a 'hit' from afrobeat the way it would from pop; it should instead focus on identifying tracks that over-perform within this niche.
