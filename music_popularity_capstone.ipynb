## ***Observation 1 (Lower Energy & Danceability, Slightly Less Acoustic)***

Compared to the global average, world-music has slightly lower acousticness (0.300 vs global 0.315), but significantly lower energy (0.533 vs global 0.641) and danceability (0.415 vs global 0.567). This means world-music is not defined by raw acoustic instrumentation, but rather by a chill, downtempo, non-aggressive sound profile. It is background listening music, not high-energy dancefloor filler. This tells the streaming service that world-music listeners are likely using it for relaxation or focus, not parties.

# Observation 2 (Totally Vocal – No Instrumental Tracks)

Critically, both the median and the 75th percentile of instrumentalness are exactly 0.000. This means every single track in the world-music dataset contains significant vocals. There is no purely instrumental folk or ceremonial music in this data. The genre is overwhelmingly driven by vocal storytelling and singing. Playlists should be organized by vocal style or language, rather than separating vocal from instrumental – because instrumental simply does not exist in this dataset.

# **Observation** 3 (The "No Hits" Ceiling) *italicized text*

The mean popularity is 41.9 out of 100, with a standard deviation of only 12.2. The 90th percentile maxes out at 55.0. This means there are zero viral mega-hits in world-music. Even the best-performing tracks barely cross the halfway mark on Spotify's popularity scale.
